```python
#hardest problem: Pyramid pattern
n=4
for i in range(0,n):
    print((n-i)*" ",end="")
    for j in range(1,i+2):
        if j==1:
            print("*",end="")
        else:
            print(" ",end="")
    for j in range(i,0,-1):
        if j==1:
            print("*",end="")
        else:
            print(" ",end="")
    print()
for i in range(n,-1,-1):
    print((n-i)*" ",end="")
    for j in range(1,i+2):
        if j==1:
            print("*",end="")
        else:
            print(" ",end="")
    for j in range(i,0,-1):
        if j==1:
            print("*",end="")
        else:
            print(" ",end="")
    print()
```
