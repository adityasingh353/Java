```python
#maxsubarraysum-kadane's algo
a=[-2, 1, -3, 4, -1, 2, 1, -5, 4]
curr=0
maxsum=float('-inf')
for i in range(len(a)):
    curr+=a[i]
    maxsum=max(curr,maxsum)
    if curr<0:
        curr=0
print("max sum subarray:",maxsum)
```
